```
import jsoncfg
from jsoncfg.config_classes import ConfigJSONObject, ConfigJSONArray, ConfigJSONScalar, ConfigNode

def recursivePrint(element):
    if isinstance(element, ConfigJSONObject):
        # Dictionary
        for key, value in element:
            print(f"key   \"{key}\" at line {jsoncfg.node_location(value).line}")
            recursivePrint(value)
    elif isinstance(element, ConfigJSONArray):
        # Array
        for item in element:
            recursivePrint(item)
    elif isinstance(element, ConfigJSONScalar):
        value = element()
        if isinstance(value, str):
            value = value.strip()

        print(f"value \"{value}\" at line {jsoncfg.node_location(element).line}")

parsed = jsoncfg.load_config(file_link)
pprint(parsed) 
type(parsed)
isinstance(parsed, ConfigNode)

# recursivePrint(parsed)
```

```
def recursivePrintPair(config_node: ConfigNode, node: dict | list | str | int | float | bool | None):
    if isinstance(config_node, ConfigJSONObject):
        # Dictionary
        assert isinstance(node, dict)
        for (config_node_key, config_node_value), (node_key, node_value) in zip(config_node, node.items()):
            print(f"key   \"{config_node_key}\" at line {jsoncfg.node_location(config_node_value).line}")
            print(f"key   \"{node_key}\" value \"{node_value}\"")
            recursivePrintPair(config_node_value, node_value)
    elif isinstance(config_node, ConfigJSONArray):
        # Array
        assert isinstance(node, list)
        for item, node_item in zip(config_node, node):
            recursivePrintPair(item, node_item)
    elif isinstance(config_node, ConfigJSONScalar):
        assert isinstance(node, str | int | float | bool | type(None))
        value = config_node()
        if isinstance(value, str):
            value = value.strip()
        print(f"value \"{value}\" at line {jsoncfg.node_location(config_node).line}")
        print(f'node value\"{node}\"')

recursivePrintPair(parsed, parsed_dict)
```

```
def print_line_numbers(element):
    if isinstance(element, dict):
        for key, value in element.items():
            # print(f"Key: '{key}' is on line {jsoncfg.node_location(element[key]).line}")
            print(f"Key: '{key}'")
            print(f"element[key] '{element[key]}'")
            print_line_numbers(value)
    elif isinstance(element, list):
        for item in element:
            print_line_numbers(item)
    elif isinstance(element, str):
        print(f"Value: '{element}' is on line {jsoncfg.node_location(element).line}")

print_line_numbers(parsed_data)
```

```
import hashlib

message = "Hello, world!"
encoded_message = message.encode('utf-8')

# Create a SHA256 hash object
hasher = hashlib.sha256()

# Update the hash object with the message
hasher.update(encoded_message)

# Get the hexadecimal digest
hex_digest = hasher.hexdigest()

print(f"SHA256 hash of '{message}': {hex_digest}")
```

```
from collections import Counter

# Counting elements in a list
data = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
fruit_counts = Counter(data)
print(f"Fruit counts: {fruit_counts}")

# Accessing counts
print(f"Count of 'apple': {fruit_counts['apple']}")
print(f"Count of 'grape': {fruit_counts['grape']}") # Returns 0 for missing key

# Finding most common elements
most_common_fruits = fruit_counts.most_common(2)
print(f"Two most common fruits: {most_common_fruits}")
```
